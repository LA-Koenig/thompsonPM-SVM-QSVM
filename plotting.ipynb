```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
from pathlib import Path
from itertools import combinations
```

```python
path = Path("./Data")
```

# PCA Data Analysis

```python
X = pd.read_csv(path / "X.csv")
X
```

```python
Y = pd.read_csv(path / "Y.csv")
Y
```

```python
Data = X.copy()
Data['Y'] = Y
Data
```

```python
# Graph the X data and use Y data for color coding
sns.scatterplot(data=Data, x="PCA1", y="PCA2", hue="Y")
plt.title("PCA of Dataset Colored by Y Values")
```

```python
fig, axes = plt.subplots(2, 3, figsize=(15, 10))
fig.suptitle("PCA Components Pairwise Scatter Plots", fontsize=16)

components = ["PCA1", "PCA2", "PCA3", "PCA4"]
for ax, (comp_x, comp_y) in zip(axes.flatten(), combinations(components, 2)):
    sns.scatterplot(data=Data, x=comp_x, y=comp_y, hue="Y", ax=ax)
    ax.set_title(f"{comp_x} vs {comp_y}")
plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```
